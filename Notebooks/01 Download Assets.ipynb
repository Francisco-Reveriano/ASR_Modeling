# Download Breeze-ASR-26 and Taiwanese-Minnan-Sutiau

Loads a Hugging Face token from `.env`, then saves the model and the dataset to local folders.

Before running, create a `.env` file in the same folder as this notebook:

```
HF_TOKEN=hf_your_token_here
```

## 1. Load environment variables from `.env`

````python
import os
from pathlib import Path

from dotenv import load_dotenv

MODEL_ID = "MediaTek-Research/Breeze-ASR-26"
DATASET_ID = "sarahwei/Taiwanese-Minnan-Sutiau"

MODEL_DIR = Path("../Models/breeze-asr-26")
DATASET_DIR = Path("../Data/taiwanese-minnan-sutiau")

load_dotenv(override=True)  # override=True picks up edits to .env without a kernel restart
hf_token = os.getenv("HF_TOKEN")
if not hf_token:
    raise ValueError("HF_TOKEN is not set. Add it to a .env file next to this notebook.")
print("HF_TOKEN loaded.")
````

## 2. Connect to Hugging Face and download the model

````python
from huggingface_hub import HfApi

user = HfApi(token=hf_token).whoami()
print(f"Connected to Hugging Face as: {user['name']}")
````

````python
from transformers import AutoModelForSpeechSeq2Seq, AutoProcessor

processor = AutoProcessor.from_pretrained(MODEL_ID, token=hf_token)
model = AutoModelForSpeechSeq2Seq.from_pretrained(MODEL_ID, device_map="auto", token=hf_token)

processor.save_pretrained(MODEL_DIR)
model.save_pretrained(MODEL_DIR)
print(f"Model saved to {MODEL_DIR.resolve()}")
````

## 3. Download the dataset

````python
from datasets import load_dataset

ds = load_dataset(DATASET_ID, token=hf_token)
ds.save_to_disk(DATASET_DIR)
print(f"Dataset saved to {DATASET_DIR.resolve()}")
ds
````
